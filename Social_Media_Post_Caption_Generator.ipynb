Social Media Post & Caption Generator

This project generates random social media captions based on user-selected categories. The dataset contains captions from influencers belonging to Beauty, Fitness, Food, and Technology categories.

```python
import pandas as pd
```

```python
df = pd.read_json("instagram-posts.json")
```

```python
all_captions = []

for inf in df.iloc[0]["influencer"]:
    all_captions.extend(inf["posts"])

print(len(all_captions))
```

```python
import random

print(random.choice(all_captions))
```

```python
all_captions = []

for inf in df.iloc[1]["influencer"]:
    all_captions.extend(inf["posts"])

print(len(all_captions))
```

```python
category = input("Enter category (Beauty, Fitness, Food, Technology): ").strip().title()

category_map = {
    "Beauty": 0,
    "Fitness": 1,
    "Food": 2,
    "Technology": 3
}

all_captions = []

for inf in df.iloc[category_map[category]]["influencer"]:
    all_captions.extend(inf["posts"])

import random

print("\nGenerated Caption:")
print("-" * 50)
print(random.choice(all_captions))
```
